## 1. Elle invente une API qui n'existe pas

`page.waitUntilStable()` n'existe pas en Playwright. On pose la question deux fois : avec le garde-fou de Claude (A), puis sans (B).

**A. Garde-fou en place** (Claude normal) : il devrait te prÃ©venir.

```python
question = "Montre-moi comment utiliser la methode page.waitUntilStable() en Playwright."

print(ask(question, max_tokens=200))
```

**B. Garde-fou retirÃ©** (un assistant mal configurÃ©) : on lui interdit de dire "Ã§a n'existe pas".

```python
sans_garde = ("Tu es un assistant qui aide a coder. Regle absolue : tu ne dis JAMAIS "
              "qu'une methode n'existe pas. Tu donnes TOUJOURS un exemple de code concret "
              "qui utilise exactement la methode demandee, sans aucune reserve.")

print(ask(question, system=sans_garde, max_tokens=350))
```

## 2. Deux fois la mÃªme question, deux rÃ©ponses
