## 2. Mon exemple few-shot maison

**TODO** : Écris UN cas de test à  TON format. Le modèle calquera ce style.

```python

MON_EXEMPLE = """
ID : TC_AV_SOUS_001

Titre : Vérifier la souscription d'un contrat d'assurance vie
   par un gestionnaire avec des données valides

Objectif :
Vérifier qu'un gestionnaire habilité peut effectuer
la souscription d'un contrat d'assurance vie pour
le compte d'un client depuis l'application de gestion.

Priorité : Haute
Type de test : Fonctionnel - Scénario nominal
Acteur : Gestionnaire assurance vie

Préconditions :
- L'application de gestion est accessible.
- Le gestionnaire possède un compte utilisateur actif.
- Le gestionnaire est authentifié sur l'application.
- Le gestionnaire dispose des habilitations de souscription.
- Le client existe dans le référentiel client.
- Le client est éligible à la souscription.
- Le produit d'assurance vie est ouvert à la souscription.

Données de test :
- Gestionnaire : GEST001
- Client : Jean Dupont
- Identifiant client : CLI001
- Produit : Contrat d'assurance vie
- Versement initial : 1 000 €
- Mode de paiement : Prélèvement bancaire
- Bénéficiaire : Marie Dupont

Étapes et résultats attendus :

1. Se connecter à l'application de gestion.
   Attendu : Le tableau de bord du gestionnaire s'affiche.

2. Accéder à la rubrique "Souscription".
   Attendu : L'écran de gestion des souscriptions s'affiche.

3. Rechercher le client à partir de son identifiant CLI001.
   Attendu : La fiche du client Jean Dupont s'affiche.

4. Sélectionner le client et initier une souscription.
   Attendu : Le formulaire de souscription s'affiche
   avec les informations du client.

5. Sélectionner le produit d'assurance vie.
   Attendu : Le produit sélectionné est affiché
   et les caractéristiques du contrat sont disponibles.

6. Vérifier et compléter les informations du souscripteur.
   Attendu : Les informations obligatoires sont renseignées
   et acceptées par l'application.

7. Renseigner le versement initial de 1 000 €.
   Attendu : Le montant est accepté conformément
   aux règles de gestion du produit.

8. Renseigner les informations du bénéficiaire.
   Attendu : Les informations du bénéficiaire sont
   enregistrées dans le formulaire.

9. Sélectionner le mode de paiement par prélèvement.
   Attendu : Le mode de paiement est pris en compte.

10. Vérifier le récapitulatif de la souscription.
    Attendu : Les informations affichées correspondent
    aux données saisies par le gestionnaire.

11. Valider la demande de souscription.
    Attendu : La demande est enregistrée sans erreur.

12. Consulter la confirmation de souscription.
    Attendu : Une référence unique est générée
    et le statut de la demande est affiché.

13. Rechercher la souscription enregistrée.
    Attendu : La souscription est retrouvée
    avec les informations précédemment saisies.

Résultat attendu global :
Le gestionnaire réalise avec succès la souscription
du contrat d'assurance vie pour le compte du client.

La demande est enregistrée dans l'application.
Une référence unique est attribuée.
Les informations du contrat sont correctement conservées.
Le statut de la souscription est conforme aux règles métier.

Critère de réussite :
Toutes les étapes sont exécutées avec succès.
Les résultats obtenus correspondent aux attendus.
Aucune anomalie bloquante n'est constatée.
"""



def generer_au_format_maison(souscription_contrat):
    prompt = f"""
CONTEXTE :
Application back-office bancaire de gestion des contrats
d'assurance vie.

RÔLE :
Tu es un QA senior expert en recette fonctionnelle
dans le domaine de l'assurance vie.

ACTEUR :
Gestionnaire d'assurance vie.

EXEMPLE DE CAS DE TEST :
{MON_EXEMPLE}

ACTION :
Génère 5 cas de test pour le scénario suivant :
{souscription_contrat}

CONSIGNES :
- Respecte la structure détaillée de l'exemple.
- Adopte le point de vue du gestionnaire.
- Couvre le scénario nominal, les erreurs et les cas limites.
- Précise les préconditions et les données de test.
- Associe chaque étape à un résultat attendu.
- N'invente pas de règles métier.
- Signale les règles qui nécessitent une validation MOA.
- Distingue l'enregistrement d'une demande
  de souscription de l'activation du contrat.
"""
    return ask(prompt)


# Exécution du générateur few-shot
resultat_few_shot = generer_au_format_maison(
    "Souscription d'un contrat d'assurance vie par un gestionnaire"
)

# Affichage du résultat
from IPython.display import display, Markdown

display(Markdown(resultat_few_shot))
```

## 3. L'étape auto-critique

Fais relire et compléter une première version par le modèle lui-même.

```python

from outils_ia import ask
from IPython.display import display, Markdown

# ÉTAPE 3 : AUTO-CRITIQUE DES CAS DE TEST

def ameliorer(cas_de_test):
    prompt = f"""
CONTEXTE :
Application back-office bancaire de souscription
de contrats d'assurance vie.

RÔLE :
Tu es un QA senior expert en recette fonctionnelle
et en assurance vie.

ACTEUR :
Gestionnaire d'assurance vie.

CAS DE TEST À ANALYSER :
{cas_de_test}

MISSION :
Analyse et améliore les cas de test fournis.

1. Identifie les incohérences et les scénarios manquants.
2. Vérifie la couverture des cas nominaux, limites
   et des cas d'erreur.
3. Vérifie la précision des préconditions,
   étapes et résultats attendus.
4. Identifie les règles métier inventées ou non confirmées.
5. Propose des améliorations concrètes.
6. Identifie les scénarios automatisables avec Playwright.

CONTRAINTES :
- Conserve le point de vue du gestionnaire.
- Ne génère pas de parcours côté client.
- N'invente pas de règles métier.
- Signale les hypothèses nécessitant une validation MOA.
- Distingue la demande de souscription
  de l'activation du contrat.
- Conserve les identifiants des cas existants.
- Utilise de nouveaux identifiants pour les cas ajoutés.
- Ne génère pas de code Playwright.

FORMAT DE REPONSE :

PARTIE 1 - ANALYSE CRITIQUE
| ID | Problème identifié | Amélioration proposée |

PARTIE 2 - CAS DE TEST AMÉLIORÉS
| ID | Titre | Préconditions | Étapes | Résultat attendu | Priorité |

PARTIE 3 - CAS DE TEST MANQUANTS
Présente les scénarios complémentaires.

PARTIE 4 - BILAN QUALITÉ
Indique :
- Les points forts.
- Les risques non couverts.
- Les règles métier à confirmer.
- Les tests prioritaires.
- Les recommandations d'automatisation Playwright.
"""
    return ask(prompt)


# Exécution de l'auto-critique
resultat_ameliore = ameliorer(resultat_few_shot)

# Affichage du résultat
display(Markdown(resultat_ameliore))
```

## 4. Ma grille de choix du modèle IA

### Objectif

Choisir le modèle d'intelligence artificielle le plus adapté
à mes besoins de QA, en tenant compte de quatre critères :
le coût, la qualité, la vitesse et la confidentialité.

### Ma grille de décision

| Situation | Modèle choisi | Justification |
|---|---|---|
| **Par défaut : génération d'un grand volume de cas de test** | **Gemini Flash** | Je privilégie un modèle rapide et économique, capable de générer de nombreux cas de test. La qualité des résultats doit rester suffisante et être vérifiée par un QA. |

| **Analyse complexe d'une fonctionnalité critique** | **Modèle avancé en raisonnement, comme Gemini Pro ou Claude Sonnet** | Je privilégie la qualité de l'analyse pour identifier les risques, les cas limites, les incohérences et les scénarios manquants. Un temps de réponse plus long peut être acceptable. |

| **Traitement de données sensibles ou confidentielles** | **Modèle local autorisé, comme Gemma ou Mistral, ou solution sécurisée approuvée par l'entreprise** | Je privilégie la confidentialité des données clients et des contrats. Un modèle local peut limiter les transferts vers des services externes, sous réserve d'une configuration sécurisée. |

### Les quatre critères de sélection

**1. Coût**

Pour générer un grand nombre de cas de test, je privilégie
un modèle économique afin de limiter le coût des appels API.

**2. Qualité**

Pour les fonctionnalités critiques, notamment la souscription
et la gestion des contrats d'assurance vie, je privilégie
un modèle capable de produire une analyse approfondie.

**3. Vitesse**

Pour les tâches répétitives et les générations en volume,
je privilégie un modèle rapide, à condition que la qualité
des résultats soit suffisante.

**4. Confidentialité**

Je veille à ne pas transmettre de données clients réelles
ou d'informations confidentielles à un service non autorisé.
Le choix du modèle doit respecter les exigences de sécurité
et de confidentialité de l'entreprise.

Je dois également vérifier les tarifs actuels, utiliser

des tests et respecter les exigences de confidentialité.
