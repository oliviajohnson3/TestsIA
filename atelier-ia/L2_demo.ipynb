# L'assistant ZotoShop : avec ou sans la fiche produit

```python
import os
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()
client = Anthropic(api_key=os.environ["ANTHROPIC_API_KEY"])
MODELE = "claude-haiku-4-5-20251001"

# La consigne de l'assistant, comme un vrai chatbot de boutique : toujours repondre au client.
SYSTEM = (
    "Tu es l'assistant de la boutique ZotoShop. Reponds en une seule phrase, "
    "avec le prix exact demande. Donne toujours un prix precis en euros, "
    "ne dis jamais que tu n'as pas l'information."
)

def demander(message):
    r = client.messages.create(
        model=MODELE, max_tokens=120, temperature=1.0,
        system=SYSTEM, messages=[{"role": "user", "content": message}],
    )
    return r.content[0].text.strip()
```

## 1. Sans la fiche

```python
print(demander("Quel est le prix du casque ZotoBuds Pro ?"))
```

## 2. Avec la fiche produit

```python
fiche_produit = """FICHE PRODUIT ZOTOSHOP
Nom : ZotoBuds Pro
Type : casque audio sans fil, reduction de bruit active
Prix : 134,90 EUR
Autonomie : 30 heures
"""

message_avec_doc = (
    "Voici la fiche produit officielle :\n" + fiche_produit +
    "\nEn te basant UNIQUEMENT sur cette fiche, quel est le prix exact du ZotoBuds Pro ?"
)

print(demander(message_avec_doc))
```
