## Une petite fonction qui mesure temps et tokens

Ici on appelle le client directement (et pas notre `ask()` de L4), parce qu'on veut **varier le modÃ¨le** : `ask()` est volontairement fixÃ© sur un seul modÃ¨le.

```python
import os, time
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()
client = Anthropic(api_key=os.environ["ANTHROPIC_API_KEY"])

def comparer(model, prompt, max_tokens=400):
    debut = time.time()
    r = client.messages.create(model=model, max_tokens=max_tokens,
                               messages=[{"role": "user", "content": prompt}])
    duree = time.time() - debut
    texte = "".join(b.text for b in r.content if b.type == "text")
    print(f"### {model}  -  {duree:.1f}s  -  {r.usage.output_tokens} tokens\n")
    print(texte)
```

## Le mÃªme prompt, Haiku vs Sonnet

```python
prompt = ("Strategie de test pour la fonction code promo du panier ZotoShop : "
          "les risques principaux et 3 cas limites non evidents.")

comparer("claude-haiku-4-5-20251001", prompt)
```

```python
comparer("claude-sonnet-4-6", prompt)
```

**Ce que tu observes** : Haiku rÃ©pond plus vite, court et net. Sonnet prend ~2x plus de temps mais structure mieux et couvre plus de cas. Le bon choix dÃ©pend de ta tÃ¢che : volume rapide â†’ Haiku ; analyse fine d'une fonction critique â†’ Sonnet.
