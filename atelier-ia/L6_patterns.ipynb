## 2. Chain-of-thought : le faire raisonner

On demande au modÃ¨le de raisonner Ã©tape par Ã©tape avant de rÃ©pondre. IdÃ©al sur les cas limites.

```python
print(ask("Quelles valeurs limites tester pour un champ quantite qui accepte de 1 a 99 ?"))
```

## 3. Auto-critique : sa propre revue

Le modÃ¨le gÃ©nÃ¨re, puis on lui fait relire et complÃ©ter son propre travail.

```python
v1 = ask("Genere 4 cas de test pour le formulaire d'inscription de ZotoShop "
         "(nom, email, mot de passe). Format tableau court.")
print(v1)
```

```python
critique = ("Voici des cas de test:\n" + v1 +
            "\n\nCritique-les : qu'est-ce qui manque cote securite, cas limites, erreurs ? "
            "Puis donne une version amelioree et completee.")

print(ask(critique))
```

## BONUS â€” Combiner les 3 patterns dans un seul prompt

Les pros ne choisissent pas : ils empilent. Un exemple (few-shot) + 
"raisonne etape par etape" (chain-of-thought) + "puis critique et ameliore" (auto-critique), 
sur un meme cas ZotoShop. C'est ce qui fait passer la sortie au niveau senior.

```python
combo = """Tu es QA senior Playwright sur ZotoShop.
US : en tant que client, je veux appliquer un code promo dans mon panier pour reduire le total.

Voici UN exemple de cas a notre format maison (suis exactement ce format) :
| ID | Titre | Preconditions | Etapes | Resultat attendu | Priorite |
| CP-01 | Code promo valide applique | panier non vide, code actif | 1. ouvrir /panier 2. saisir le code 3. valider | total reduit, remise affichee | P1 |

1) Raisonne d'abord etape par etape : classes d'equivalence du code promo 
(valide / expire / inexistant / deja utilise / non cumulable), puis valeurs limites, puis cas.
2) Genere les cas au format ci-dessus.
3) Critique ta propre liste : qu'est-ce qui manque cote erreurs, cas limites, securite ? 
Donne enfin une version amelioree et completee."""

print(ask(combo, max_tokens=1500))
```
