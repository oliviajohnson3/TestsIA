## Temps 2 - L'appel devient un test qui se verifie lui-meme

J'emballe `ask()` dans `generer_cas` et un mini-nettoyeur `parse_json` (plomberie), puis les 3 `assert` verifient le travail de l'IA tout seuls : JSON valide, au moins 5 cas, le bon format. C'est le reflexe 'l'IA propose, tu valides' (L8) devenu du code.

````python
def parse_json(texte):
    """L'IA enrobe parfois son JSON de balises ```json ou de texte.
    On garde juste du premier [ au dernier ] avant de parser."""
    debut, fin = texte.find('['), texte.rfind(']')
    return json.loads(texte[debut:fin + 1])

def generer_cas(feature):
    return ask(
        f'Genere 5 cas de test pour "{feature}" de ZotoShop. '
        'Reponds UNIQUEMENT en JSON, sans markdown : '
        'une liste d\'objets {"id", "titre", "priorite"}.'
    )

def test_sortie_exploitable():
    cas = parse_json(generer_cas('page de connexion'))
    assert isinstance(cas, list)           # JSON valide : une liste
    assert len(cas) >= 5                    # au moins 5 cas
    assert all('titre' in c for c in cas)   # le bon format
    print('OK :', len(cas), 'cas verifies')

test_sortie_exploitable()
````

## Temps 3 - Un seul run, un fichier de cas

Je boucle sur plusieurs fonctionnalites, je rassemble tous les cas et je les ecris dans un fichier CSV (ouvrable dans Excel). ChatGPT aussi peut sortir du JSON ou un fichier â€” la vraie difference n'est pas la. Elle est ici : ce code tourne tout seul, dans ton pipeline, sur 20 features comme sur 200, avec le prompt et la verification versionnes dans Git. Le chat, c'est toi qui pilotes a la main, une fois. Le code, c'est une etape automatisable de ton systeme (on le pousse en CI/CD au module 4).

````python
import csv

tous_les_cas = []
for fonctionnalite in ['page de connexion', 'panier', 'paiement']:
    for cas in parse_json(generer_cas(fonctionnalite)):
        cas['fonctionnalite'] = fonctionnalite
        tous_les_cas.append(cas)

# On ecrit le batch dans un fichier exploitable (impossible depuis le chat)
with open('cas_de_test_zotoshop.csv', 'w', newline='', encoding='utf-8') as fichier:
    writer = csv.DictWriter(
        fichier,
        fieldnames=['fonctionnalite', 'id', 'titre', 'priorite'],
        extrasaction='ignore')
    writer.writeheader()
    writer.writerows(tous_les_cas)

print(len(tous_les_cas), 'cas generes en un seul run -> cas_de_test_zotoshop.csv')
for cas in tous_les_cas[:3]:
    print(' -', cas.get('id'), cas.get('titre'))
````

> **Piege JSON** : l'IA enrobe parfois sa sortie de balises ```json. Notre `parse_json` (defini au Temps 2) garde juste du premier `[` au dernier `]` â€” nettoyer la sortie d'une IA fait partie du job.
