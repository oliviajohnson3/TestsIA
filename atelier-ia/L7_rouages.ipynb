## On appelle le client directement

Ici on lit les compteurs de tokens, que notre `ask()` n'expose pas. Donc on passe par le client Anthropic.

```python
import os
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()
client = Anthropic(api_key=os.environ["ANTHROPIC_API_KEY"])
MODEL = "claude-haiku-4-5-20251001"
```

## 1. Compter les tokens : ce que tu paies

Avant mÃªme d'envoyer, on peut compter les tokens d'entrÃ©e d'un prompt.

```python
texte = "Genere trois cas de test pour la page panier de ZotoShop."

compte = client.messages.count_tokens(model=MODEL,
                                      messages=[{"role": "user", "content": texte}])
print("Tokens d'entree :", compte.input_tokens)
```

Et aprÃ¨s un vrai appel, le compteur d'usage donne l'entrÃ©e ET la sortie (les deux sont facturÃ©es).

```python
r = client.messages.create(model=MODEL, max_tokens=300,
                           messages=[{"role": "user", "content": texte}])
print("Entree :", r.usage.input_tokens, "tokens")
print("Sortie :", r.usage.output_tokens, "tokens")
print("Total facture :", r.usage.input_tokens + r.usage.output_tokens, "tokens")
```

## 2. La rÃ©ponse coupÃ©e : max_tokens

`max_tokens` plafonne la longueur de la rÃ©ponse. Trop bas = rÃ©ponse tronquÃ©e.

```python
r = client.messages.create(model=MODEL, max_tokens=20,
                           messages=[{"role": "user",
                                      "content": "Explique en detail la strategie de test d'un panier e-commerce."}])

print("stop_reason :", r.stop_reason)
print("---")
print("".join(b.text for b in r.content if b.type == "text"))
```
